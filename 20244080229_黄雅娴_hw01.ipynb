欧几里得算法

```python
def gcd(m, n):
    m,n =abs(m),abs(n)
    while n:
        m,n=n,m%n
    return m
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))

print(gcd(-12,8))
print(gcd(12,-8))
print(gcd(-12,-8))
```

该函数通过在入口处对参数取绝对值，确保无论输入是否包含负数，始终返回符合数学定义的非负最大公约数；这一行为既避免了 Python 取模运算在负数场景下可能导致的符号异常，又与标准库 math.gcd() 的语义保持一致，在分数化简、密码学等实际应用中保证了结果的正确性与通用性，因此是合理且必要的标准实现方式。

近似欧拉数e

```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    if n <= 0:
        return 0.0
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
    return total

def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    factorial = 1
    k = 0
    while True:
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
        if abs(total - true_e) < epsilon:
            break
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print(euler_limit(1000))
print(euler_infinite_sum(0))
print(euler_infinite_sum(1))
print(euler_infinite_sum(4))
print(euler_approx(1e-6))
print_euler_sum_table(10)
```

测试整数属性

```python
def is_power_of_2(n):
    if n == 1:
        return True
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return False

def is_power(b, n):
    if n == 1:
        return True
    if b == 1:
        return True
    current = 1
    while current < b:
        current *= n
        if current == b:
            return True
    return False

print(is_power_of_2(1))
print(is_power_of_2(2))
print(is_power_of_2(8))
print(is_power_of_2(16))
print(is_power_of_2(3))
print(is_power_of_2(10))

print(is_power(8, 2))
print(is_power(27, 3))
print(is_power(1, 5))
print(is_power(10, 2))
print(is_power(64, 4))
print(is_power(100, 10))
```
