实验2：函数、条件、循环与递归

```python
def score_level(score):
    if not isinstance(score, (int, float)):
        raise ValueError("分数必须是数值类型")
    if score < 0 or score > 100:
        raise ValueError("分数必须在0到100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"
total_for = 0
for i in range(1, 101):
    total_for += i
total_while = 0
n = 1
while n <= 100:
    total_while += n
    n += 1
print(f"for循环结果: {total_for}")
print(f"while循环结果: {total_while}")
print(f"两种实现结果是否一致: {total_for == total_while}")
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
for val in [0, 1, 5, 10]:
    rec = factorial_recursive(val)
    ite = factorial_iterative(val)
    print(f"n={val}: 递归={rec}, 迭代={ite}, 一致={rec == ite}")
def summary(values):
    min_val = values[0]
    max_val = values[0]
    total = 0
    for v in values:
        if v < min_val:
            min_val = v
        if v > max_val:
            max_val = v
        total += v
    avg_val = total / len(values)
    return (min_val, max_val, avg_val)
test_list = [3, 1, 4, 1, 5, 9, 2, 6]
minimum, maximum, average = summary(test_list)
print(f"最小值: {minimum}, 最大值: {maximum}, 平均值: {average}")
def sum_all(*args):
    total = 0
    for arg in args:
        total += arg
    return total
print(f"无参数: {sum_all()}")
print(f"单个参数: {sum_all(42)}")
print(f"多个参数: {sum_all(1, 2, 3, 4, 5)}")
```

```python
import os
import pickle

cwd = os.getcwd()
print(f"当前工作目录: {cwd}")
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"数据目录已就绪: {data_dir}")

file_path = os.path.join(data_dir, "students.txt")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""

with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
print(f"\n文件共 {len(lines)} 行(含表头)")
for line in lines[1:]:
    print(f"  原始行: {line.strip()}")
def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量错误(期望2个,实际{len(fields)}个)")
    name, score_str = fields[0].strip(), fields[1].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩'{score_str}'无法转换为整数")
    if score < 0 or score > 100:
        raise ValueError(f"成绩{score}超出0~100范围")
    return {"name": name, "score": score}
valid_records = []
invalid_records = []
for idx, line in enumerate(lines[1:], start=2):
    stripped = line.strip()
    if not stripped:
        continue
    try:
        record = parse_student(stripped)
        valid_records.append(record)
    except ValueError as e:
        invalid_records.append({"line": idx, "content": stripped, "error": str(e)})

print(f"\n有效记录: {len(valid_records)} 条")
for r in valid_records:
    print(f"  {r}")
print(f"无效记录: {len(invalid_records)} 条")
for r in invalid_records:
    print(f"  第{r['line']}行 '{r['content']}' -> {r['error']}")

report_path = os.path.join(data_dir, "report.txt")
avg_score = sum(r["score"] for r in valid_records) / len(valid_records)
pass_count = sum(1 for r in valid_records if r["score"] >= 60)
pass_rate = pass_count / len(valid_records)

with open(report_path, "w", encoding="utf-8") as f:
    f.write("=== 成绩统计报告(w模式写入) ===\n")
    f.write(f"有效记录数: {len(valid_records)}\n")
    f.write(f"无效记录数: {len(invalid_records)}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")
print(f"\n[w模式] report.txt 已写入(覆盖)")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
with open(report_path, "a", encoding="utf-8") as f:
    f.write("\n=== 补充说明(a模式追加) ===\n")
    f.write("以上数据仅包含有效记录。\n")
    f.write("无效记录已单独记录,未纳入统计。\n")
print("[a模式] report.txt 已追加内容")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)
print(f"pickle恢复数据与原数据相等: {loaded_records==valid_records}")
print(f"恢复的记录: {loaded_records}")
module_path = os.path.join(cwd, "student_utils.py")
with open(module_path, "w", encoding="utf-8") as f:
    f.write("def parse_student(line):\n")
    f.write("fields=line.strip().split(',')\n")
    f.write("if len(fields) != 2:\n")
    f.write("raise ValueError(f'字段数量错误(期望2个,实际{len(fields)}个)')\n")
    f.write("name,score_str=fields[0].strip(), fields[1].strip()\n")
    f.write("if not name:\n")
    f.write("raise ValueError('姓名为空')\n")
    f.write("try:\n")
    f.write("score=int(score_str)\n")
    f.write("except ValueError:\n")
    f.write("raise ValueError(f\"成绩'{score_str}'无法转换为整数\")\n")
    f.write("if score < 0 or score > 100:\n")
    f.write("raise ValueError(f'成绩{score}超出0~100范围')\n")
    f.write("return {'name': name, 'score': score}\n")

import importlib
import student_utils
importlib.reload(student_utils)

print("\n--- 模块导入测试 ---")
try:
    result = student_utils.parse_student("Alice,85")
    print(f"有效输入 'Alice,85' -> {result}")
except ValueError as e:
    print(f"错误: {e}")

try:
    result = student_utils.parse_student("Grace,abc")
    print(f"无效输入 'Grace,abc' -> {result}")
except ValueError as e:
    print(f"无效输入 'Grace,abc' -> ValueError: {e}")

nonexistent_path = os.path.join(data_dir, "not_exist.txt")
try:
    with open(nonexistent_path, "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError:
    print(f"\nFileNotFoundError: 文件 '{nonexistent_path}' 不存在,请检查路径是否正确。")
```

实验5：类、继承与对象复制

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
valid_records = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92},
    {"name": "David", "score": 76},
    {"name": "Eva", "score": 60},
    {"name": "Frank", "score": 89},
]
students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students:
    print(f"{s} | 及格: {s.is_passed()}")
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")
gs = GraduateStudent("Alice", 85, "NLP")
print(gs)
print(f"及格: {gs.is_passed()}")
original = {"name": "Alice", "scores": [85, 90]}

direct = original
direct["scores"].append(95)
print(f"直接赋值后 - original: {original}")
print(f"直接赋值后 - direct:   {direct}")
print(f"同一对象: {original is direct}")
print()

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"].append(95)
print(f"浅拷贝后 - original: {original}")
print(f"浅拷贝后 - shallow:  {shallow}")
print(f"内部列表同一对象: {original['scores'] is shallow['scores']}")
print()

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"].append(95)
print(f"深拷贝后 - original: {original}")
print(f"深拷贝后 - deep:     {deep}")
print(f"内部列表同一对象: {original['scores'] is deep['scores']}")
```
