Python 正则表达式练习

```python
import re

# 1. 检查字符串是否仅包含英文大小写字母和数字
def check_alphanumeric(s):
    return bool(re.fullmatch(r'[a-zA-Z0-9]+', s))

# 2. 匹配一个 a 后面跟着零个或多个 b
def match_a_followed_by_bs(s):
    return bool(re.fullmatch(r'ab*', s))

# 3. 匹配一个 a 后面跟着一个或多个 b
def match_a_followed_by_one_or_more_bs(s):
    return bool(re.fullmatch(r'ab+', s))

# 4. 匹配以 a 开头、以 b 结尾的字符串
def match_a_start_b_end(s):
    return bool(re.fullmatch(r'a.*b', s))

# 5. 删除 IP 地址中各段数字的前导零
def remove_leading_zeros_ip(ip):
    return re.sub(r'\b0+(\d)', r'\1', ip)

# 6. 将空格替换为下划线，并将下划线替换为空格
def swap_spaces_and_underscores(s):
    # 使用 \s 匹配空白，或者使用 r'[ _]' 和字典替换
    return re.sub(r'[ _]', lambda m: '_' if m.group() == ' ' else ' ', s)

# 7. 将 yyyy-mm-dd 转换为 dd-mm-yyyy
def convert_date_format(date_str):
    return re.sub(r'(\d{4})-(\d{2})-(\d{2})', r'\3-\2-\1', date_str)

# 8. 找出所有以 a 或 e 开头的单词
def find_words_starting_with_a_or_e(s):
    return re.findall(r'\b[ae]\w*', s, re.IGNORECASE)

# 9. 将 Road 缩写为 Rd
def abbreviate_road(s):
    return re.sub(r'\bRoad\b', 'Rd', s)

# 10. 找出所有长度至少为 4 个字符的单词
def find_long_words(s):
    return re.findall(r'\b\w{4,}\b', s)

# 11. 删除除字母和数字之外的所有字符
def remove_non_alphanumeric(s):
    return re.sub(r'[^a-zA-Z0-9]', '', s)

# 12. 删除括号及括号内的内容
def remove_brackets_content(s):
    return re.sub(r'\s*\([^)]*\)', '', s)


# ================= 测试用例 =================
if __name__ == "__main__":
    print("1. 仅包含字母数字:", check_alphanumeric("Hello123"), check_alphanumeric("Hello 123"))
    print("2. a后零或多个b:", match_a_followed_by_bs("a"), match_a_followed_by_bs("abbb"), match_a_followed_by_bs("ac"))
    print("3. a后一个或多个b:", match_a_followed_by_one_or_more_bs("ab"), match_a_followed_by_one_or_more_bs("a"))
    print("4. a开头b结尾:", match_a_start_b_end("a123b"), match_a_start_b_end("ab"))

    print("5. IP去前导零:", remove_leading_zeros_ip("192.168.001.010"), remove_leading_zeros_ip("010.000.000.001"))
    print("6. 空格下划线互换:", swap_spaces_and_underscores("hello_world test"))

    print("7. 日期格式转换:", convert_date_format("2024-05-01"))
    print("8. a或e开头的单词:", find_words_starting_with_a_or_e("An apple a day keeps the doctor away, every evening."))
    print("9. Road缩写:", abbreviate_road("Main Road, Crossroad, Roadway"))

    print("10. 长度>=4的单词:", find_long_words("This is a simple test string with words."))
    print("11. 去除非字母数字:", remove_non_alphanumeric("Hello, World! 123 - @#$"))

    print("12. 删除括号及内容:")
    texts = ["example (.com)", "w3resource", "github (.com)", "stackoverflow (.com)"]
    for t in texts:
        print(f"   原文本: {t:25} -> 处理后: {remove_brackets_content(t)}")
```

```python
from bs4 import BeautifulSoup
import re
mock_html = """
<!DOCTYPE html>
<html>
<head><title>Mock Climate Page</title></head>
<body>
<h2>Climate</h2>
<table class="wikitable">
    <tbody>
        <tr>
            <th colspan="14">Climate data for Harbin (1991–2020 normals)</th>
        </tr>
        <tr>
            <th scope="col">Month</th>
            <th scope="col">Jan</th>
            <th scope="col">Feb</th>
            <th scope="col">Mar</th>
        </tr>
        <tr>
            <th scope="row">Record high °C (°F)</th>
            <td>4.2<br>(39.6)</td>
            <td>9.9<br>(49.8)</td>
            <td>22.6<br>(72.7)</td>
        </tr>
        <tr>
            <th scope="row">Mean daily maximum °C (°F)</th>
            <td>-12.3<br>(9.9)</td>
            <td>-7.5<br>(18.5)</td>
            <td>2.0<br>(35.6)</td>
        </tr>
        <tr>
            <th scope="row">Average precipitation mm (inches)</th>
            <td>4.2<br>(0.17)</td>
            <td>5.6<br>(0.22)</td>
            <td>13.7<br>(0.54)</td>
        </tr>
        <tr>
            <td colspan="14" style="text-align:center;">Source: China Meteorological Administration</td>
        </tr>
    </tbody>
</table>
</body>
</html>
"""

def get_soup_from_url(url):
    if url == "mock://harbin":
        return BeautifulSoup(mock_html, "html.parser")
    return None

# ================= 任务1：读取并解析 =================
print("=== 任务1 ===")
url_harbin = "mock://harbin"
soup = get_soup_from_url(url_harbin)
climate_th = None
for th in soup.find_all("th"):
    text = re.sub(r"\s+", " ", th.get_text()).strip()
    if text.startswith("Climate data for"):
        climate_th = th
        break

table = climate_th.find_parent("table") if climate_th else None
if table:
    direct_children = [c.name for c in table.children if c.name]
    print("找到表格: True")
    print("表格直接子节点标签:", direct_children)
    print("直接子节点数量:", len(direct_children))

# ================= 任务2：retrieve_climate_table =================
def retrieve_climate_table(s):
    soup = get_soup_from_url(s)
    if not soup:
        return None
    for th in soup.find_all("th"):
        text = re.sub(r"\s+", " ", th.get_text()).strip()
        if text.startswith("Climate data for"):
            tbl = th.find_parent("table")
            if tbl:
                return tbl
    for cap in soup.find_all("caption"):
        text = re.sub(r"\s+", " ", cap.get_text()).strip()
        if text.startswith("Climate data for"):
            tbl = cap.find_parent("table")
            if tbl:
                return tbl

    return None

# ================= 任务3：list_climate_table_row_names =================
def list_climate_table_row_names(s):
    table = retrieve_climate_table(s)
    if table is None:
        return None

    row_names = []
    for tr in table.find_all("tr"):
        row_th = tr.find("th", attrs={"scope": "row"})

        if row_th is None:
            ths = tr.find_all("th")
            if not ths:
                continue
            first_th_text = re.sub(r"\s+", " ", ths[0].get_text()).strip()
            if first_th_text.lower() == "month":
                continue
            if first_th_text.startswith("Climate data for"):
                continue
            if len(ths) > 3:
                continue
            row_th = ths[0]

        if row_th is None:
            continue

        name = re.sub(r"\s+", " ", row_th.get_text()).strip()
        if not name or name.startswith("Climate data for"):
            continue
        if name.lower().startswith("source"):
            continue
        row_names.append(name)

    return row_names

# ================= 测试 =================
print("\n=== 任务2 测试 ===")
tbl = retrieve_climate_table("mock://harbin")
print("找到表格对象:", type(tbl).__name__ if tbl else None)

print("\n=== 任务3 测试 ===")
names = list_climate_table_row_names("mock://harbin")
print("提取的行名:")
for n in names:
    print(" -", n)
```
