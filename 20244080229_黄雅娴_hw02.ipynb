字符串的乐趣

```python
def is_palindrome(s):
    cleaned = ''.join(c.lower() for c in s if c != ' ')
    return cleaned == cleaned[::-1]

def is_abecedarian(s):
    cleaned = ''.join(c for c in s if c != ' ')
    return cleaned == ''.join(sorted(cleaned))

def remove_vowels(s):
    vowels = 'aeiou'
    return ''.join(c for c in s if c.lower() not in vowels)

print("=== 测试 is_palindrome ===")
print(is_palindrome("level"))
print(is_palindrome("kayak"))
print(is_palindrome("pop"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("Was it a car or a cat I saw"))
print(is_palindrome("tacocat"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))
print(is_palindrome("world"))
print("\n=== 测试 is_abecedarian ===")
print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
print(is_abecedarian("zyx"))
print("\n=== 测试 remove_vowels ===")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
print(remove_vowels("AEIOU"))
print(remove_vowels("xyz"))
print(remove_vowels("Hello World"))
```

列表的乐趣

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("Input must be a list")
    return lst[::-1]

def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print("=== 测试 list_reverse ===")
print(list_reverse([1, 2, 3]))
print(list_reverse([1]))
print(list_reverse([]))
print(list_reverse(['a', 'b', 'c']))

try:
    list_reverse("hello")
except TypeError as e:
    print("捕获 TypeError:", e)

try:
    list_reverse(123)
except TypeError as e:
    print("捕获 TypeError:", e)

print("\n=== 测试 is_sorted ===")
print(is_sorted([1, 2, 3, 4, 5]))
print(is_sorted([1, 2, 2, 3]))
print(is_sorted([5, 4, 3, 2, 1]))
print(is_sorted([1, 3, 2]))
print(is_sorted([]))
print(is_sorted([1]))
print(is_sorted("abcde"))
print(is_sorted("aceb"))
print(is_sorted((1, 2, 3)))
```

作为向量的元组

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("First argument must be a tuple")
    if not isinstance(s, (int, float)):
        raise TypeError("Second argument must be an int or float")
    for item in t:
        if not isinstance(item, (int, float)):
            raise TypeError("All elements in the tuple must be int or float")
    return tuple(item * s for item in t)


def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("Both arguments must be tuples")
    if len(t1) != len(t2):
        raise ValueError("Tuples must have the same length")
    for item in t1:
        if not isinstance(item, (int, float)):
            raise TypeError("All elements in the first tuple must be int or float")
    for item in t2:
        if not isinstance(item, (int, float)):
            raise TypeError("All elements in the second tuple must be int or float")
    return float(sum(a * b for a, b in zip(t1, t2)))


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return False
    first_row_len = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if first_row_len is None:
            first_row_len = len(row)
        elif len(row) != first_row_len:
            return False
        for item in row:
            if not isinstance(item, (int, float)):
                return False
    return True


def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise TypeError("First argument must be a valid matrix (tuple of tuples of numbers)")
    if not isinstance(v, tuple):
        raise TypeError("Second argument must be a tuple")
    for item in v:
        if not isinstance(item, (int, float)):
            raise TypeError("All elements in the vector must be int or float")
    if len(mx[0]) != len(v):
        raise ValueError("Matrix columns and vector length must match")
    result = []
    for row in mx:
        row_sum = 0.0
        for a, b in zip(row, v):
            row_sum += a * b
        result.append(row_sum)
    return tuple(result)


print("=== 测试 vec_scalar_mult ===")
print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.0, 2.5), 3))
try:
    vec_scalar_mult([1, 2, 3], 2)
except TypeError as e:
    print("捕获 TypeError:", e)
try:
    vec_scalar_mult((1, 2, 3), "a")
except TypeError as e:
    print("捕获 TypeError:", e)

print("\n=== 测试 vec_inner_product ===")
print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1.0, 2.0), (3.0, 4.0)))
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("捕获 ValueError:", e)
try:
    vec_inner_product([1, 2], (1, 2))
except TypeError as e:
    print("捕获 TypeError:", e)

print("\n=== 测试 check_valid_mx ===")
print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2, 3),)))
print(check_valid_mx(((1, 2), (3,))))
print(check_valid_mx(((1, 2), [3, 4])))
print(check_valid_mx(()))

print("\n=== 测试 mx_vec_mult ===")
print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
print(mx_vec_mult(((1, 0, 0), (0, 1, 0), (0, 0, 1)), (1, 2, 3)))
try:
    mx_vec_mult(((1, 2), (3, 4)), (1, 2, 3))
except ValueError as e:
    print("捕获 ValueError:", e)
```

更多向量的乐趣

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or dimension < 0:
            raise ValueError("Dimension must be a non-negative integer")
        self.dimension = dimension
        if elements is None:
            self.elements = tuple([0.0] * dimension)
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("Elements must be a list or tuple")
            if len(elements) != dimension:
                raise ValueError("Number of elements must match dimension")
            for e in elements:
                if not isinstance(e, (int, float)):
                    raise TypeError("All elements must be int or float")
            self.elements = tuple(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("Cannot compare vectors of different dimensions")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("Cannot compare vectors of different dimensions")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("Cannot compare vectors of different dimensions")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("Cannot compare vectors of different dimensions")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("Cannot compare vectors of different dimensions")
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("Argument must be a Vector")
        if self.dimension != other.dimension:
            raise ValueError("Dimensions must match for dot product")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dimension, tuple(e * other for e in self.elements))
        if isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("Dimensions must match for element-wise multiplication")
            return Vector(self.dimension, tuple(a * b for a, b in zip(self.elements, other.elements)))
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"


v1 = Vector(3, (1, 2, 3))
v2 = Vector(3, (1, 2, 3))
v3 = Vector(3, (1, 2, 4))
v4 = Vector(3, (4, 5, 6))

print("v1 =", v1)
print("v2 =", v2)
print("v3 =", v3)
print("v4 =", v4)

print("\n=== 比较运算 ===")
print("v1 == v2:", v1 == v2)
print("v1 == v3:", v1 == v3)
print("v1 < v3:", v1 < v3)
print("v1 <= v3:", v1 <= v3)
print("v3 > v1:", v3 > v1)
print("v3 >= v1:", v3 >= v1)

try:
    v1 < Vector(2, (1, 2))
except ValueError as e:
    print("捕获 ValueError:", e)

print("\n=== 内积 ===")
print("v1.dot(v4):", v1.dot(v4))

print("\n=== 乘法运算 ===")
print("v1 * 2:", v1 * 2)
print("2 * v1:", 2 * v1)
print("v1 * v4:", v1 * v4)

try:
    v1 * Vector(2, (1, 2))
except ValueError as e:
    print("捕获 ValueError:", e)
```
