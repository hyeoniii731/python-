实验1：正则表达式的基础匹配

```python
import re

text1 = "cat on mat; later: cats and dogs"

print("=== 任务1 ===")
print("匹配数字 \\d+:", re.findall(r"\d+", text1))
print("匹配空白 \\s+:", re.findall(r"\s+", text1))
print("匹配单词 cat \\bcat\\b:", re.findall(r"\bcat\b", text1))
print("\n=== 任务2 ===")
text2 = "cat dog bird CAT"
print("cat|dog 匹配:", re.findall(r"cat|dog", text2))

print("a|aa|aaa on 'aaaa':", re.findall(r"a|aa|aaa", "aaaa"))
print("aaa|aa|a on 'aaaa':", re.findall(r"aaa|aa|a", "aaaa"))
print("\n=== 任务3 ===")
text3 = "ct cat caat caaat"

patterns = [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]
for p in patterns:
    print(f"{p:10} -> {re.findall(p, text3)}")
```

实验2：分组、回溯引⽤与替换

```python
import re

print("=== 任务1：捕获组提取邮箱用户名和域名 ===")
text1 = "联系：alice@example.com；备选：bob@test.org"
result1 = re.findall(r"([\w.]+)@([\w.]+)", text1)
for user, domain in result1:
    print(f"用户名: {user}, 域名: {domain}")

print("\n=== 任务2：后向引用找连续重复单词 ===")
text2 = "cat cat, dog cat, bird bird"
result2 = re.findall(r"\b(\w+)\s+\1\b", text2)
print("连续重复的单词:", result2)

print("\n=== 任务3：日期格式转换 年-月-日 -> 月-日-年 ===")
text3 = "2024-05-01\n2024/05/02\n2024.05.03"
print("转换前:")
print(text3)
converted = re.sub(r"(\d{4})([-./])(\d{2})\2(\d{2})", r"\3\2\1", text3)
converted = re.sub(r"(\d{4})([-./])(\d{2})\2(\d{2})", lambda m: f"{m.group(3)}{m.group(2)}{m.group(4)}{m.group(2)}{m.group(1)}", text3)
print("\n转换后:")
print(converted)

print("\n=== 任务4：电话号码统一格式 ===")
text4 = "联系我：123-456-7890 或 987.654.3210"
result4 = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", text4)
print("转换后:", result4)
```

实验3：使⽤ BeautifulSoup 解析 HTML

```python
!pip install beautifulsoup4 -i https://pypi.tuna.tsinghua.edu.cn/simple
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

print("=== 任务1：标题与h1 ===")
print("页面标题文本:", soup.title.get_text())
h1 = soup.h1
print("h1标签名:", h1.name)
print("h1文本:", h1.get_text())

print("\n=== 任务2：find_all 查找 ===")
courses = soup.find_all("p", class_="course")
links = soup.find_all("a")
print("class=course 的段落:")
for p in courses:
    print("  ", p.get_text())
print("所有链接:")
for a in links:
    print("  文字:", a.get_text(), "| href:", a.get("href"))

print("\n=== 任务3：get_text 比较 ===")
print("整个文档可见文本:")
print(repr(soup.get_text()))
print("\n单个段落文本:", repr(courses[0].get_text()))
print("单个链接文本:", repr(links[0].get_text()))
```
